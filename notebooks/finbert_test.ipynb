```python
from transformers import pipeline

finbert = pipeline("text-classification", model="ProsusAI/finbert", top_k=None)

headlines = [
    "Apple beats earnings expectations, raises dividend",
    "JPMorgan shares fall after weak trading revenue",
    "Microsoft to hold annual shareholder meeting in December",
]

for headline, scores in zip(headlines, finbert(headlines)):
    print(headline, {s["label"]: round(s["score"], 3) for s in scores})
```
